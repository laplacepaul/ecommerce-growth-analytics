```python
# ============================================
# Import Required Libraries
# ============================================

from pathlib import Path

import pandas as pd
```

```python
# ============================================
# Configure Processed Data Path
# ============================================

PROJECT_ROOT = Path.cwd().parent
PROCESSED_PATH = PROJECT_ROOT / "data" / "processed"

print("Processed path:", PROCESSED_PATH.resolve())
print("Processed path exists:", PROCESSED_PATH.exists())
```

```python
# ============================================
# Inspect Processed Warehouse Schemas
# ============================================

warehouse_files = [
    "fact_orders",
    "dim_customer",
    "dim_product",
    "dim_seller",
    "dim_review"
]

warehouse_schema = {}

for table_name in warehouse_files:
    file_path = PROCESSED_PATH / f"{table_name}.csv"

    warehouse_schema[table_name] = pd.read_csv(
        file_path,
        nrows=5
    )

    print(f"\n{table_name}")
    print("-" * len(table_name))
    print(warehouse_schema[table_name].dtypes)
```

```python
# ============================================
# Build Warehouse Column Inventory
# ============================================

column_inventory = pd.concat(
    [
        pd.DataFrame({
            "table_name": table_name,
            "column_name": dataframe.columns,
            "pandas_type": dataframe.dtypes.astype(str).values
        })
        for table_name, dataframe in warehouse_schema.items()
    ],
    ignore_index=True
)

column_inventory
```
